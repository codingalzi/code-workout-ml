## 머신러닝 모델 훈련의 어려움

머신러닝 모델을 성공적으로 학습하기 어려운 이유는 크게 **데이터의 문제**와 **모델 학습의 문제**로 나눌 수 있다.

**데이터 문제**

- **훈련 데이터가 충분하지 않음**: 머신러닝 모델이 안정적으로 학습하려면 문제에 따라 많은 데이터가 필요할 수 있다.
- **훈련 데이터의 품질이 낮음**: 편향된 데이터, 잘못 측정된 값, 오류가 포함된 데이터는 모델의 학습과 예측에 나쁜 영향을 줄 수 있다.
- **적절한 특성을 찾기 어려움**: 문제 해결에 도움이 되는 특성을 선택하거나 새롭게 구성하는 **특성 공학**이 필요할 수 있다.

**모델 학습의 문제**

- **과대적합**: 모델이 훈련 데이터에 지나치게 맞춰져 새로운 데이터에서는 성능이 떨어지는 현상이다.
- **과소적합**: 모델이 데이터에 나타난 관계를 충분히 학습하지 못해 훈련 데이터에서도 좋은 성능을 내지 못하는 현상이다.

과대적합과 과소적합은 모델뿐 아니라 데이터의 양과 품질, 사용한 특성 등 여러 요인의 영향을 받을 수 있다.

### 선형 회귀 모델

1인당 GDP가 높을수록 삶의 만족도도 대체로 높아지는 경향을 하나의 직선으로 표현해 보자.
이를 위해 1인당 GDP를 입력으로 받아 삶의 만족도를 예측하는 **선형 회귀 모델**(linear regression model)을 사용한다.

가장 단순한 형태의 선형 회귀 모델은 다음과 같은 일차식으로 표현할 수 있다.

$$
\widehat{\text{삶의 만족도}}
=
\theta_0
+
\theta_1 \times \text{1인당 GDP}
$$

여기서 $\theta_0$는 **절편**, $\theta_1$은 **기울기**이며, 이 두 값을 모델의 **파라미터**(parameter)라고 한다.
같은 GDP가 주어져도 $\theta_0$와 $\theta_1$의 값에 따라 예측값이 달라지므로, 어떤 파라미터를 선택하느냐가 모델의 예측 성능을 결정한다.

아래 그림은 서로 다른 절편과 기울기를 갖는 세 개의 직선을 비교한다.
어떤 직선은 데이터의 전체적인 증가 경향을 어느 정도 반영하지만, 어떤 직선은 그렇지 못한다.
따라서 선형 회귀에서는 **데이터에 가장 잘 맞는 절편과 기울기를 찾는 과정**이 중요하다.

각 국가에는 실제 삶의 만족도 값이 있고, 하나의 직선을 정하면 각 GDP에 대한 예측값을 계산할 수 있다.
모델 학습은 실제값과 예측값의 차이가 전체적으로 작아지도록 파라미터를 정하는 과정으로 이해할 수 있다.

머신러닝 모델은 훈련 데이터에 지나치게 맞춰질 경우 새로운 데이터에서 충분한 성능을 내지 못할 수 있다.
이를 **과대적합**(overfitting)이라고 한다.

여기서는 훈련 데이터의 구성이 달라질 때 선형 회귀 모델이 어떻게 달라지는지 살펴본다.

9개 국가를 포함했을 때와 제외했을 때 학습된 선형 회귀 모델은 상당히 다르다.
이 예는 **훈련 데이터의 구성에 따라 학습된 모델이 크게 달라질 수 있으며, 새로운 데이터에서의 성능을 따로 확인해야 한다**는 점을 보여준다.

특히 모델이 훈련 데이터에 지나치게 맞춰져 새로운 데이터에서 성능이 떨어지는 현상을 **과대적합**(overfitting)이라고 한다.

### 모델 규제

선형 회귀를 포함한 모든 머신러닝 모델은 훈련 과정에서 과대적합될 수 있다.
모델이 훈련 데이터의 세부적인 특성이나 잡음에 지나치게 민감하게 반응하면 새로운 데이터에 대한 예측 성능이 떨어질 수 있다.

이를 완화하기 위해 모델의 복잡도나 파라미터의 크기를 제한하는 **규제**(regularization)를 사용할 수 있다.
선형 회귀에 규제를 적용한 대표적인 모델이 **릿지 회귀**(Ridge regression)이다.

아래 그래프에서 파란색 파선은 9개 국가를 제외한 데이터로 학습한 릿지 회귀 모델이고, 검은색 실선은 전체 국가로 학습한 선형 회귀 모델이다.
두 직선의 차이가 크지 않은 것은 규제가 훈련 데이터의 변화에 대한 모델의 민감도를 줄이는 데 도움이 될 수 있음을 보여준다.

<div align="center">
    <img src="https://github.com/codingalzi/code-workout-ml/blob/master/images/ch01/homl01-21-c.png?raw=true" width="650">
</div>

규제를 적용하면 훈련 데이터에 대한 성능은 다소 낮아질 수 있지만, 특정 훈련 데이터에 지나치게 맞춰지는 것을 줄일 수 있다.
그 결과 새로운 데이터에 대한 예측 성능인 **일반화 성능**(generalization performance)이 좋아질 수 있다.

성공적인 머신러닝 모델을 만들기 위해서는 다음 두 가지를 함께 고려해야 한다.

1. 훈련 데이터에서 충분히 작은 예측 오차를 얻는다.
2. 과대적합을 줄여 새로운 데이터에서도 좋은 성능을 유지한다.
