머신러닝은 데이터에서 **변수 사이의 관계나 데이터의 구조를 학습**하고, 이를 새로운 데이터에 대한 **예측이나 판단**에 활용하는 방법이다.

## 머신러닝은 무엇이 다른가?

### 전통적인 프로그래밍

전통적인 프로그래밍에서는 사람이 문제를 분석하고 해결 규칙을 직접 코드로 작성한다.

예를 들어 스팸 메일을 분류하려면 “제목에 특정 단어가 포함되면 스팸으로 분류한다”와 같은 규칙을 사람이 정할 수 있다.

새로운 스팸 표현이 계속 등장하면 규칙도 계속 추가하거나 수정해야 한다. 문제의 규칙이 많고 자주 변할수록 프로그램은 복잡해지고 유지하기 어려워진다.

전통적인 프로그래밍은 일정한 과정을 따라 진행된다.
먼저 문제 연구 단계에서 해결해야 할 문제를 분석하고, 이를 해결할 수 있는 알고리즘을 연구한다. 
다음으로 규칙 작성 단계에서는 연구된 알고리즘을 실제 코드로 구현한다. 
이후에는 평가 단계가 이어지는데, 이 단계에서는 구현된 프로그램을 테스트하여 올바르게 작동하는지 확인한다.
테스트를 통과하면 프로그램은 실전 배치된다.
그러나 테스트에 실패할 경우, 발생한 오차를 분석한 뒤 다시 문제 연구 단계로 돌아가 알고리즘을 수정하고 개선하는 과정을 반복한다.

### 머신러닝

머신러닝에서는 사람이 모든 규칙을 직접 작성하지 않는다.

스팸 메일과 정상 메일의 예를 충분히 제공하면, 모델이 두 집단을 구분하는 데 도움이 되는 관계를 데이터에서 학습한다.

> **데이터 → 모델 학습 → 새로운 데이터에 대한 예측**

즉 머신러닝의 핵심은 규칙을 사람이 일일이 작성하는 대신, **데이터를 이용해 예측에 필요한 관계를 학습하는 것**이다.

스팸 메일 분류 머신러닝 모델의 학습은 스팸 메일과 아닌 메일의 구분법을 학습하기 위해
다량의 스팸 메일과 스팸이 아닌 메일로 구성된 훈련셋을 활용한다.
학습이 완료된 스팸 메일 분류 머신러닝 모델은 새로운 이메일에 대해
학습된 정보를 이용하여 스팸인지 아닌지 판별한다.

머신러닝은 다음과 같은 문제에서 특히 유용하다.

- 사람이 명시적인 규칙을 모두 작성하기 어려운 문제
- 규칙이 자주 변하는 문제
- 많은 데이터에서 예측에 필요한 관계를 찾아야 하는 문제

음성 인식, 이미지 인식, 추천, 스팸 분류, 수요 예측 등이 대표적인 예다.

## 머신러닝 문제의 기본 유형

이 강의에서는 먼저 **지도 학습**과 **비지도 학습**을 구분한다.

| 유형 | 설명 | 대표 과제 |
|---|---|---|
| 지도 학습 | 입력 데이터와 타깃을 함께 사용 | 회귀, 분류 |
| 비지도 학습 | 타깃 없이 입력 데이터만 사용 | 군집화, 차원 축소, 이상치 탐지 |

지도 학습에서는 타깃의 종류에 따라 다시 나눌 수 있다.

- **회귀**(regression): 수치 값을 예측
- **분류**(classification): 범주를 예측

이 장에서는 회귀 문제를 하나 직접 따라가면서 모델이 실제로 무엇을 학습하는지 살펴본다.

## 1인당 GDP로 삶의 만족도를 예측할 수 있을까?

OECD 회원국과 일부 비회원국의 **1인당 국내총생산**(GDP per capita)과 **삶의 만족도**(Life Satisfaction) 사이의 관계를 살펴보자.

질문은 단순하다.

> **한 나라의 1인당 GDP를 알면 그 나라의 삶의 만족도를 예측할 수 있을까?**

이 문제에서

- **특성**(feature): 1인당 GDP
- **타깃**(target): 삶의 만족도

이다.

삶의 만족도는 수치이므로 이 문제는 **회귀 문제**다.

아래 표는 데이터셋에 포함된 7개 국가의 1인당 GDP와 삶의 만족도를 예제로 보여준다.
1인당 GDP는 미국 달러를 기준으로 계산되었으며, 삶의 만족도는 0부터 10 사이의 소수점을 갖는다.

산점도를 보면 1인당 GDP가 높을수록 삶의 만족도도 대체로 높아지는 **선형적**(linear) 경향이 보인다.

그렇다면 이 관계를 하나의 직선으로 표현해 볼 수 있을까?

## 선형 회귀 모델

가장 단순한 선형 회귀 모델은 다음과 같이 표현할 수 있다.

$$
\widehat{y} = \theta_0 + \theta_1 x
$$

여기서

- $x$: 1인당 GDP
- $\widehat{y}$: 모델이 예측한 삶의 만족도
- $\theta_0$: **절편**
- $\theta_1$: **기울기**

이다.

$\theta_0$와 $\theta_1$을 모델의 **파라미터**(parameter)라고 한다.

같은 GDP가 주어져도 파라미터 값이 달라지면 예측값도 달라진다. 따라서 선형 회귀 모델을 학습한다는 것은

> **훈련 데이터에 적절한 절편과 기울기를 찾는 것**

이라고 이해할 수 있다.

<div align="center">
    <img src="https://github.com/codingalzi/code-workout-ml/blob/master/images/ch01/homl01-18-b.png?raw=true" width="450">
</div>

### 어떤 직선이 좋은가?

같은 데이터에도 여러 직선을 그을 수 있다.

어떤 직선은 데이터의 전체적인 증가 경향을 잘 반영하지만, 어떤 직선은 그렇지 못한다.

각 국가에는 실제 삶의 만족도 값이 있고, 하나의 직선을 정하면 각 GDP에 대한 예측값이 생긴다. 따라서 각 국가마다 **실제값과 예측값의 차이**도 생긴다.

선형 회귀 모델의 학습은 이러한 차이가 전체적으로 작아지도록 파라미터를 정하는 과정이다.

구체적인 계산 방법은 뒤의 모델 훈련 장에서 다룬다. 지금은 다음만 기억하면 충분하다.

> **학습 = 데이터로부터 적절한 모델 파라미터를 찾는 과정**

<div align="center">
    <img src="https://github.com/codingalzi/code-workout-ml/blob/master/images/ch01/homl01-19-b.png?raw=true" width="450">
</div>

학습된 모델은 새로운 국가의 1인당 GDP가 주어졌을 때 삶의 만족도를 예측하는 데 사용할 수 있다.

즉,

- **훈련**(training): 데이터에서 파라미터를 정한다.
- **예측**(prediction): 학습된 파라미터를 새로운 입력에 적용한다.

머신러닝의 목적은 훈련 데이터를 다시 계산하는 것이 아니라, **새로운 데이터에서도 적절한 예측을 하는 것**이다.

두 경우에 학습된 직선은 상당히 다르다.

이 예는 중요한 사실을 보여준다.

> **모델은 데이터에서 학습하므로 어떤 데이터를 사용했는가에 따라 학습 결과가 달라질 수 있다.**

따라서 훈련 데이터에 잘 맞는 모델을 얻었다는 사실만으로 충분하지 않다. 실제로 중요한 것은 **학습에 사용하지 않은 새로운 데이터에서도 잘 작동하는가**이다.

이를 **일반화**(generalization)라고 한다.

## 과대적합과 과소적합

### 과대적합

**과대적합**(overfitting)은 모델이 훈련 데이터에 지나치게 맞춰져 새로운 데이터에서 성능이 떨어지는 현상이다.

훈련 데이터의 세부적인 특성이나 잡음까지 지나치게 반영하면 이런 문제가 생길 수 있다.

### 과소적합

**과소적합**(underfitting)은 모델이 지나치게 단순하거나 학습이 충분하지 않아 데이터에 나타난 관계를 제대로 표현하지 못하는 현상이다.

따라서 좋은 모델은 단순히 훈련 데이터를 가장 잘 맞히는 모델이 아니다.

> **새로운 데이터에서도 안정적으로 작동하는 모델이 좋은 모델이다.**

## 과대적합을 줄이는 한 가지 방법: 규제

과대적합을 줄이기 위해 모델이 지나치게 복잡해지지 않도록 제한할 수 있다. 이를 **규제**(regularization)라고 한다.

선형 회귀에 규제를 적용한 대표적인 모델이 **릿지 회귀**(Ridge regression)다.

규제를 사용하면 훈련 데이터에 대한 성능은 조금 낮아질 수 있지만, 특정 훈련 데이터에 지나치게 민감하게 반응하는 것을 줄여 **일반화 성능**을 높이는 데 도움이 될 수 있다.

## 모델만큼 데이터도 중요하다

머신러닝의 성능은 알고리즘에만 달려 있지 않다.

대표적인 데이터 문제는 다음과 같다.

- **훈련 데이터가 충분하지 않음**: 안정적인 관계를 학습하기 어렵다.
- **훈련 데이터의 품질이 낮음**: 잘못 측정된 값이나 오류가 모델의 학습을 방해한다.
- **훈련 데이터가 대표적이지 않음**: 실제 사용할 상황과 다른 데이터로 학습하면 일반화하기 어렵다.
- **적절한 특성을 찾기 어려움**: 문제 해결에 도움이 되는 특성을 선택하거나 새롭게 구성해야 할 수 있다.

이후의 프로젝트에서는 모델을 고르기 전에 먼저 **데이터가 어떤 문제를 가지고 있는지** 확인한다.

## 이 장의 핵심

이 장의 GDP–삶의 만족도 예제에는 이후 계속 사용할 머신러닝의 핵심 개념이 들어 있다.

1. **특성과 타깃을 정한다.**
2. **모델의 형태를 정한다.**
3. **훈련 데이터로 모델의 파라미터를 학습한다.**
4. **학습된 모델로 새로운 데이터를 예측한다.**
5. **훈련 데이터가 달라지면 학습된 모델도 달라질 수 있다.**
6. 따라서 훈련 성능보다 **일반화 성능**이 중요하다.
7. 과대적합을 줄이기 위해 규제와 같은 방법을 사용할 수 있다.

한 문장으로 정리하면 다음과 같다.

> **머신러닝의 목표는 훈련 데이터를 가장 잘 맞히는 것이 아니라, 데이터에서 유용한 관계를 학습해 새로운 데이터에서도 잘 작동하는 모델을 만드는 것이다.**

### 문제 2. GDP와 삶의 만족도 선형 회귀

1인당 GDP를 이용하여 삶의 만족도를 예측하는 다음 선형 회귀 모델을 생각하자.

$$
\widehat{y} = \theta_0 + \theta_1 x
$$

<details>
<summary><strong>해설 보기</strong></summary>
